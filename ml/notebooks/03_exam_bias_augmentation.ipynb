```python
import os
import cv2
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
def desk_crop(img):
    h, w, _ = img.shape
    return img[int(h * 0.25):h, :]

def random_occlusion(img, size=120):
    h, w, _ = img.shape
    x = random.randint(0, w - size)
    y = random.randint(0, h - size)
    img[y:y+size, x:x+size] = 0
    return img
```

```python
img = cv2.imread("../datasets/yolo_exam/images/train/" + random.choice(
    os.listdir("../datasets/yolo_exam/images/train")
))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

aug = desk_crop(img.copy())
aug = random_occlusion(aug)

plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.title("Exam-biased")
plt.imshow(aug)
plt.axis("off")
```
